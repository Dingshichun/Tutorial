# Anaconda 使用

### （1）安装和配置

下载安装之后需要配置好环境变量，然后才能在命令行使用相关命令，比如conda。在环境变量中添加自己 Anaconda 的路径，我的路径如下。在安装的过程中也会有将 Anaconda 路径添加到环境变量的选项，勾选即可，就不用手动配置。
```
D:\SoftwareInstall\Anaconda
D:\SoftwareInstall\Anaconda\Scripts
D:\SoftwareInstall\Anaconda\Library\mingw-w64\bin
D:\SoftwareInstall\Anaconda\Library\usr\bin
D:\SoftwareInstall\Anaconda\Library\bin
```
一些简单的 conda 命令如下，在终端输入运行：
```
# 注意，"#" 后面的内容是注释，不输入终端。
# 1.环境命令
conda --version # 查看 conda 版本
conda info --envs # 列出创建的所有环境
conda env list # 列出创建的所有环境
conda create --name 环境名 python=3.9 # 创建环境并指定 python版本
conda activate 环境名 # 激活环境
conda deactivate # 退出当前环境
conda env remove --name 环境名 # 删除环境
conda env export > environment.yaml # 导出环境配置到文件，注意保存文件的路径不要有中文，会生成失败。
conda env create -f environment.yaml # 从配置文件创建环境

# 2. 包命令
conda install 包名 # 安装包，可自动解决依赖
conda install 包名=版本号 # 安装指定版本的包
conda remove 包名 # 卸载包
conda update 包名 # 更新包
conda update --all # 更新所有包
conda list # 列出当前环境中所有的包
conda search 包名 # 搜索可安装的包版本

3.配置和工具
conda config --show # 查看当前配置
conda config --add channels 镜像地址 # 添加镜像源（如清华源）
conda clean --all # 清理缓存和临时文件
conda info # 查看 conda 系统信息
```
简单介绍一下 pip 和 conda 的区别。他们都是 Python 生态中常用的​​包管理工具​​，但在设计目标、依赖管理和适用场景上有显著区别。conda 相当于一个全能管家，pip 则适合纯 python 开发，conda 能自动解析依赖树，在 conda 环境中优先使用 conda ，必要时补充 pip，优先使用 conda 安装基础包，pip 补充。

**根据文件中指定的版本来安装包**  
假如有文件 requirements.txt ，其指定了 numpy 和 matplot 的版本，如下：
```
matplotlib>=3.3
numpy>=1.23.5
```
就可以直接根据该文件来下载对应的包，首先在 requirements.txt 文件所在的文件夹打开终端，激活要安装这些包的环境，然后输入对应的安装指令即可，如果已经有对应的版本文件，安装会很方便。如下所示。需要注意的是，torch 和 torchvision 是需要单独下载的，不适合以这种方式下载。
```
# 假如现在已经在打开的终端里
activate yolov5 # 激活环境
pip install -r requirements.txt # 安装命令
```
